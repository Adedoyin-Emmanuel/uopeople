```python
# =================================================================
# PROGRAMMING ASSIGNMENT: FLOWER SPECIES PREDICTION LAB CODE
# =================================================================

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.datasets import load_iris
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.metrics import classification_report, confusion_matrix

# Machine Learning Classifier Dependencies
from sklearn.tree import DecisionTreeClassifier
from sklearn.naive_bayes import GaussianNB
from sklearn.linear_model import LogisticRegression
from sklearn.neighbors import KNeighborsClassifier
from sklearn.svm import SVC

# -------------------------------------------------------------
# STEP 1: DATA CONFIGURATION & METRIC VISUALIZATION PREPARATION
# -------------------------------------------------------------

# Load data array structures directly from sklearn
iris_raw_data = load_iris()
X_features = iris_raw_data.data
y_targets = iris_raw_data.target
feature_labels = iris_raw_data.feature_names
target_species = iris_raw_data.target_names

# Map attributes directly into a structured DataFrame format
iris_dataframe = pd.DataFrame(X_features, columns=feature_labels)
iris_dataframe['species'] = y_targets

print("Program Check: Dataset features and labels successfully mapped.")
print(f"Total metrics rows: {iris_dataframe.shape[0]} | Attributes counted: {iris_dataframe.shape[1]-1}")

# Split features into 75% training allocations and 25% test allocations
X_train, X_test, y_train, y_test = train_test_split(
    X_features, y_targets, test_size=0.25, random_state=42, stratify=y_targets
)

# Standardize values uniformly across distance and optimization frameworks
data_scaler = StandardScaler()
X_train_scaled = data_scaler.fit_transform(X_train)
X_test_scaled = data_scaler.transform(X_test)


# Modular evaluation script to cleanly print metrics and plot matrices
def compute_and_plot_metrics(model_name, y_true, y_pred, is_rule_based=False):
    print(f"\n" + "="*50)
    print(f" PERFORMANCE METRICS: {model_name}")
    print("="*50)
    
    # Calculate confusion matrix structure
    cm_structure = confusion_matrix(y_true, y_pred)
    print("\nCalculated Confusion Matrix Array:")
    print(cm_structure)
    
    # Print out summary records for Precision, Recall, and F1-Score
    print("\nDetailed Metric Categorization Summary:")
    print(classification_report(y_true, y_pred, target_names=target_species))
    
    # Plot matrix via seaborn heatmap to capture output visual screenshots
    plt.figure(figsize=(5.5, 4))
    sns.heatmap(cm_structure, annot=True, fmt='d', cmap='Blues', cbar=False,
                xticklabels=target_species, yticklabels=target_species)
    plt.title(f'Confusion Matrix: {model_name}')
    plt.xlabel('Predicted Structural Labels')
    plt.ylabel('Actual Biological Labels')
    plt.tight_layout()
    plt.show()


# -------------------------------------------------------------
# STEP 2A: BUILDING TREE-BASED & PROBABILISTIC EXPERIMENTS
# -------------------------------------------------------------

# Experiment 1: Standard Decision Tree Architecture
tree_classifier = DecisionTreeClassifier(random_state=42)
tree_classifier.fit(X_train, y_train)
tree_outcomes = tree_classifier.predict(X_test)
compute_and_plot_metrics("Decision Tree Framework", y_test, tree_outcomes)


# Experiment 2: Explicit Structural If-Else Logic System
def execute_custom_rule_model(data_matrix):
    predicted_records = []
    for measurement in data_matrix:
        petal_length_value = measurement[2]
        petal_width_value = measurement[3]
        
        # Isolate target groupings using distinct structural cutoffs
        if petal_length_value < 2.5 and petal_width_value < 0.8:
            predicted_records.append(0)  # Categorized as Setosa
        elif petal_length_value < 4.8 and petal_width_value < 1.6:
            predicted_records.append(1)  # Categorized as Versicolor
        else:
            predicted_records.append(2)  # Categorized as Virginica
    return np.array(predicted_records)

rule_outcomes = execute_custom_rule_model(X_test)
compute_and_plot_metrics("Custom Structural Rule System", y_test, rule_outcomes, is_rule_based=True)


# Experiment 3: Probabilistic Gaussian Naïve Bayes
bayes_classifier = GaussianNB()
bayes_classifier.fit(X_train, y_train)
bayes_outcomes = bayes_classifier.predict(X_test)
compute_and_plot_metrics("Gaussian Naïve Bayes Model", y_test, bayes_outcomes)


# -------------------------------------------------------------
# STEP 2B: BUILDING DISTANCE & OPTIMIZATION-BASED EXPERIMENTS
# -------------------------------------------------------------

# Experiment 4: Multiclass Optimizing Logistic Regression Model
logistic_classifier = LogisticRegression(random_state=42, max_iter=200)
logistic_classifier.fit(X_train_scaled, y_train)
logistic_outcomes = logistic_classifier.predict(X_test_scaled)
compute_and_plot_metrics("Logistic Regression Framework", y_test, logistic_outcomes)


# Experiment 5: Distance-Based Spatial K-Nearest Neighbors
knn_spatial_model = KNeighborsClassifier(n_neighbors=5, metric='minkowski', p=2)
knn_spatial_model.fit(X_train_scaled, y_train)
knn_spatial_outcomes = knn_spatial_model.predict(X_test_scaled)
compute_and_plot_metrics("K-Nearest Neighbors (KNN)", y_test, knn_spatial_outcomes)


# Experiment 6: Optimization Hyperplane Support Vector Machine
svm_hyperplane_model = SVC(kernel='rbf', random_state=42)
svm_hyperplane_model.fit(X_train_scaled, y_train)
svm_hyperplane_outcomes = svm_hyperplane_model.predict(X_test_scaled)
compute_and_plot_metrics("Support Vector Machine (SVM)", y_test, svm_hyperplane_outcomes)
```
